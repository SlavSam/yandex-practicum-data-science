График тренда показывает стабильное увеличение среднего, начиная с апреля 2018 г.

График сезонности указывает на еженедельную циклично повторяющуюся закономерность, что логично в свете анализа количества заказов такси.

На графике остатков декомпозиции наблюдаются хаотичные выбросы разной амплитуды. Вероятно, это связано со стабильным ростом тренда.

Создадим новые признаки: календарные, отстающие значения и скользящее среднее. Объявим функцию для создания признаков.

```python
def make_features(data, max_lag, rolling_mean_size):
    """
    Создает новые признаки для временного ряда.

    Параметры:
    data (DataFrame): Входные данные в формате pandas DataFrame,
                      индекс которого является временной меткой.
    max_lag (int): Максимальный лаг для создания лаговых признаков.
    rolling_mean_size (int): Размер окна для вычисления скользящего среднего.

    Возвращает:
    DataFrame: Обновленный DataFrame с добавленными признаками (год, месяц, день, день недели, лаговые признаки и скользящее среднее).
    """

    # Извлечение временных характеристик из индекса
    data['dayofweek'] = data.index.dayofweek  # День недели (0=понедельник, 6=воскресенье)

    # Создание лаговых признаков
    for lag in range(1, max_lag + 1):
        data['lag_{}'.format(lag)] = data['num_orders'].shift(lag)  # Лаговый признак с задержкой на lag периодов

    # Вычисление скользящего среднего и его сдвиг
    data['rolling_mean_{}'.format(rolling_mean_size)] = data['num_orders'].rolling(rolling_mean_size).mean().shift()

    return data


df = make_features(df, 24, 24)
display(df.head(30))
```

## Обучение моделей

Обучим и протестируем на кросс-валидации несколько моделей: линейную регрессию, случайный лес, нейросеть RNN (Recurrent Neural Networks), модель классического статистического анализа SARIMA (Seasonal ARIMA), а также нейросеть `torch.nn`.

Чтобы делать кросс-валидацию на основе метрики RMSE для моделей временных рядов, нужно учитывать, что стандартные методы кросс-валидации, такие как KFold, не подходят. Вместо этого мы будем использовать временную кросс-валидацию с помощью `TimeSeriesSplit` и затем вычислим RMSE для каждого периода.

Но, сначала с помощью подстановки предыдущего значения создадим случайные предсказания для проверки указанных выше моделей на адекватность.

```python
# Оценка производительности модели с использованием RMSE
rmse = np.sqrt(mean_squared_error(test['num_orders'], test['lag_1']))
rmse = round(rmse, 2)
print(f'Среднеквадратическое отклонение предсказаний на основе lag_1 (RMSE): {rmse}\n')
```

### Модель `LinearRegression`


```python
X = train.drop(columns='num_orders')  # Признаки
y = train['num_orders']               # Целевая переменная

X_test = test.drop(columns='num_orders')
y_test = test['num_orders']
```

```python
num_columns = ['lag_1', 'lag_2', 'lag_23', 'lag_24', 'rolling_mean_24']

# Временная кросс-валидация
tscv = TimeSeriesSplit(n_splits=5)

# Линейная регрессия с временной кросс-валидацией и метрикой RMSE
model_lr = LinearRegression()
rmse_scores_lr = []

# Кросс-валидация для линейной регрессии
for train_index, test_index in tscv.split(X):
    X_train_split, X_test_split = X.iloc[train_index], X.iloc[test_index]
    y_train_split, y_test_split = y.iloc[train_index], y.iloc[test_index]

    # Создание пайплайна для предобработки данных
    preprocessor = ColumnTransformer(
        transformers=[
            ('num', Pipeline([
                ('imputer', SimpleImputer(strategy='median')),
                ('scaler', MinMaxScaler())
            ]), num_columns),  # Обязательно определите num_columns заранее
        ],
        remainder='passthrough'  # Оставляем остальные столбцы без изменений
    )

    # Создание полного пайплайна
    pipeline_lr = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('model', model_lr)
    ])

    # Обучение модели
    pipeline_lr.fit(X_train_split, y_train_split)

    # Предсказание
    y_pred_lr = pipeline_lr.predict(X_test_split)

    # Вычисление RMSE
    rmse_lr = np.sqrt(mean_squared_error(y_test_split, y_pred_lr))
    rmse_scores_lr.append(rmse_lr)

print(f'RMSE для линейной регрессии: {round(np.mean(rmse_scores_lr), 2)}')
```

```python
# Функция для формирования входа (X) и выхода (y) для LSTM
def create_dataset(X, y, time_steps=1):
    """
    Формирует датасет для LSTM, используя указанные временные шаги.

    Параметры:
    X (pd.DataFrame): Входные данные.
    y (pd.Series): Выходные данные.
    time_steps (int): Количество временных шагов для создания входных данных.

    Возвращает:
    tuple: Массивы входных данных (Xs) и выходных данных (ys).
    """
    Xs, ys = [], []
    for i in range(len(X) - time_steps):
        Xs.append(X.iloc[i:(i + time_steps)].values)  # Индексация входных данных
        ys.append(y.iloc[i + time_steps])  # Индексация выходных данных
    return np.array(Xs), np.array(ys)

# Задаем количество временных шагов
time_steps = 5  # Измените на нужное значение

# Создаем финальные выборки
X_final, y_final = create_dataset(X, y, time_steps)

# Просматриваем размеры получившихся массивов
print("Размер X:", X_final.shape)
print("Размер y:", y_final.shape)
```

```python
# Класс-обертка для Keras
class KerasRegressorCustom(BaseEstimator, RegressorMixin):
    def __init__(self, units=32, dropout_rate=0.0, batch_size=10, epochs=100):
        """
        Инициализация параметров модели Keras.

        Параметры:
        - units (int): Количество единиц LSTM в каждом слое.
        - dropout_rate (float): Доля выхода, которую следует сбрасывать (dropout).
        - batch_size (int): Размер пакета для обучения.
        - epochs (int): Количество эпох для обучения.
        """
        self.units = units
        self.dropout_rate = dropout_rate
        self.batch_size = batch_size
        self.epochs = epochs
        self.model = None

    def create_model(self, input_shape):
        """
        Создание модели LSTM.

        Параметры:
        - input_shape (tuple): Форма входных данных.

        Возвращает:
        - model (Sequential): Скомпилированная модель Keras.
        """
        model = Sequential()
        model.add(Input(shape=input_shape))
        model.add(LSTM(units=self.units, return_sequences=True))
        model.add(Dropout(self.dropout_rate))
        model.add(LSTM(units=self.units))
        model.add(Dropout(self.dropout_rate))
        model.add(Dense(1))
        model.compile(optimizer='adam', loss='mean_squared_error')
        return model

    def fit(self, X, y):
        """
        Обучение модели.

        Параметры:
        - X (pd.DataFrame): Входные данные.
        - y (pd.Series): Выходные данные.
        """
        X_processed = preprocessor.fit_transform(X)  # Обработка данных
        self.input_shape = (X_processed.shape[1], 1)  # Устанавливаем форму входных данных
        self.model = self.create_model(self.input_shape)
        self.model.fit(X_processed, y, batch_size=self.batch_size, epochs=self.epochs, verbose=1)

    def predict(self, X):
        """
        Предсказание на новых данных.

        Параметры:
        - X (pd.DataFrame): Входные данные.

        Возвращает:
        - numpy.ndarray: Предсказанные значения.
        """
        X_processed = preprocessor.transform(X)  # Применение предобработки
        return self.model.predict(X_processed)

# Параметры поиска
param_grid = {
    'model__units': [16],
    'model__dropout_rate': [0.0],
    'model__batch_size': [10, 20],
    'model__epochs': [10, 20]
}

# Создание экземпляра KerasRegressor
pipeline_keras = Pipeline(steps=[
    ('model', KerasRegressorCustom())
])

# Запуск подбора гиперпараметров
grid_search_keras = GridSearchCV(estimator=pipeline_keras, param_grid=param_grid, cv=tscv,
                                  scoring='neg_mean_squared_error', n_jobs=-1)

# Обучение подбора гиперпараметров
grid_search_keras.fit(X, y)

# Получение лучших параметров
best_params_keras = grid_search_keras.best_params_
print(f'Лучшие параметры: {best_params_keras}')

# Получение средних оценок RMSE для каждого набора параметров
mean_rmse_scores = np.sqrt(-grid_search_keras.cv_results_['mean_test_score'])
print(f'Средний RMSE на кросс-валидации: {round(np.mean(mean_rmse_scores), 2)}')

# Получение наилучшей модели
best_model_keras = grid_search_keras.best_estimator_
```

### Модель SARIMA

```python
# Определение пользовательского класса для модели SARIMA
class SARIMAPipeline(BaseEstimator, RegressorMixin):
    def __init__(self, order=(0, 0, 0), seasonal_order=(0, 0, 0, 1)):
        """
        Инициализация параметров модели SARIMA.

        Параметры:
        order (tuple): Порядок модели SARIMA (p, d, q).
        seasonal_order (tuple): Порядок сезонной модели SARIMA (P, D, Q, s).
        """
        self.order = order
        self.seasonal_order = seasonal_order
        self.model = None

    def fit(self, y):
        """
        Обучение модели SARIMA.

        Параметры:
        y (pd.Series): Временной ряд для обучения модели.
        """
        self.model = SARIMAX(y, order=self.order, seasonal_order=self.seasonal_order)
        self.model_fit = self.model.fit(disp=False)
        return self

    def predict(self, steps):
        """
        Прогнозирование на заданное количество шагов вперед.

        Параметры:
        steps (int): Количество шагов вперед для прогнозирования.

        Возвращает:
        np.ndarray: Прогнозируемые значения.
        """
        return self.model_fit.forecast(steps=steps)

    def cross_val_score(self, y, n_splits=5):
        """
        Оценка модели с использованием кросс-валидации.

        Параметры:
        y (pd.Series): Временной ряд для оценки модели.
        n_splits (int): Количество разбиений для кросс-валидации.

        Возвращает:
        float: Среднее значение RMSE по всем сплитам.
        """
        tscv = TimeSeriesSplit(n_splits=n_splits)  # Инициализация временной кросс-валидации
        rmse_scores = []  # Список для хранения значений RMSE

        for train_index, test_index in tscv.split(y):
            y_train, y_test = y.iloc[train_index], y.iloc[test_index]  # Обучающая и тестовая выборки

            self.fit(y_train)  # Обучение модели
            y_pred = self.predict(len(y_test))  # Прогнозирование

            rmse = np.sqrt(mean_squared_error(y_test, y_pred))  # Расчет RMSE
            rmse_scores.append(rmse)

        return np.mean(rmse_scores)  # Возврат среднего RMSE по всем сплитам

# Определение диапазонов гиперпараметров для поиска
p_values = range(0, 2)  # Значения для p: порядок авторегрессии
d_values = range(0, 2)  # Значения для d: порядок интегрирования
q_values = range(0, 2)  # Значения для q: порядок скользящего среднего
P_values = range(0, 2)  # Значения для P: порядок сезонной авторегрессии
D_values = range(0, 1)  # Значения для D: порядок сезонного интегрирования
Q_values = range(0, 1)  # Значения для Q: порядок сезонного скользящего среднего
s = 24  # Длина сезона (например, 24 для часовых данных)

# Создание параметров сетки для поиска
param_grid = ParameterGrid({
    'order': [(p, d, q) for p in p_values for d in d_values for q in q_values],
    'seasonal_order': [(P, D, Q, s) for P in P_values for D in D_values for Q in Q_values]
})

# Подбор гиперпараметров
best_rmse = float('inf')  # Инициализация лучшего значения RMSE
best_order = None  # Инициализация лучших параметров order
best_seasonal_order = None  # Инициализация лучших сезонных параметров

# Перебор параметров
for params in param_grid:
    order = params['order']  # Получение текущих параметров order
    seasonal_order = params['seasonal_order']  # Получение текущих сезонных параметров

    try:
        # Создание экземпляра SARIMAPipeline
        sarima_model = SARIMAPipeline(order=order, seasonal_order=seasonal_order)

        # Выполнение кросс-валидации
        rmse = sarima_model.cross_val_score(y)

        # Проверка на наилучший RMSE
        if rmse < best_rmse:
            best_rmse = rmse
            best_order = order
            best_seasonal_order = seasonal_order

    except Exception as e:
        print(f"Ошибка для параметров {order} и {seasonal_order}: {e}")

# Вывод лучших параметров и RMSE
print(f'Лучший порядок: {best_order}, Лучший сезонный порядок: {best_seasonal_order}, RMSE: {best_rmse}')
```

### Нейросеть torch.nn


```python
# Определение класса нейронной сети
class Net(nn.Module):
    def __init__(self, n_in_neurons, n_hidden_neurons_1, n_hidden_neurons_2, n_out_neurons):
        super(Net, self).__init__()
        self.fc1 = nn.Linear(n_in_neurons, n_hidden_neurons_1)
        self.act1 = nn.ReLU()
        self.fc2 = nn.Linear(n_hidden_neurons_1, n_hidden_neurons_2)
        self.act2 = nn.ReLU()
        self.fc3 = nn.Linear(n_hidden_neurons_2, n_out_neurons)
        self._initialize_weights()

    def _initialize_weights(self):
        nn.init.kaiming_normal_(self.fc1.weight, mode='fan_in')
        nn.init.normal_(self.fc1.bias, mean=0.5, std=0.7)
        nn.init.kaiming_normal_(self.fc2.weight, mode='fan_in')
        nn.init.normal_(self.fc2.bias, mean=0.5, std=0.7)
        nn.init.kaiming_normal_(self.fc3.weight, mode='fan_in')
        nn.init.normal_(self.fc3.bias, mean=0.5, std=0.7)

    def forward(self, x):
        x = self.fc1(x)
        x = self.act1(x)
        x = self.fc2(x)
        x = self.act2(x)
        x = self.fc3(x)
        return x
```

```python
def fit(
    net,
    optimizer,
    criterion,
    X_train,
    y_train,
    X_val,
    y_val,
    train_loss,  # Список для сохранения потерь
    test_metric,  # Список для сохранения метрики
    num_epochs=10000,
    patience=100,
    max_delta=1
):
    """
    Обучение модели нейронной сети с использованием ранней остановки.

    Параметры:
    - net: Модель нейронной сети.
    - optimizer: Оптимизатор для обновления весов.
    - criterion: Функция потерь.
    - X_train: Входные данные для обучения.
    - y_train: Целевые метки для обучения.
    - X_val: Входные данные для валидации.
    - y_val: Целевые метки для валидации.
    - train_loss: Список для сохранения потерь обучения.
    - test_metric: Список для сохранения метрик валидации.
    - num_epochs: Общее количество эпох (по умолчанию 10000).
    - patience: Параметр для ранней остановки (число эпох без улучшения).
    - max_delta: Максимально допустимое изменение RMSE для остановки.

    Возвращает:
    - train_loss: Обновленный список потерь обучения.
    - test_metric: Обновленный список метрик валидации.
    """

    net.train()
    min_test_metric = np.inf
    counter = 0

    for epoch in range(num_epochs):
        optimizer.zero_grad()  # Обнуляем градиенты
        preds = net(X_train)   # Прогноз на обучающей выборке

        # Расчет потерь
        loss_value = criterion(preds, y_train)
        loss_value.backward()  # Обратное распространение ошибки
        optimizer.step()       # Обновление весов

        # Мониторинг потерь и метрик
        if epoch % 10 == 0 or epoch == num_epochs - 1:
            train_loss.append(torch.sqrt(loss_value).detach().cpu().numpy())

            net.eval()  # Переключаем модель в режим оценки
            with torch.no_grad():
                val_preds = net(X_val)
                rmse = torch.sqrt(criterion(val_preds, y_val)).detach().cpu().numpy()
                test_metric.append(rmse)  # Добавляем значение RMSE в список метрик валидации

            # Ранняя остановка
            if rmse < min_test_metric:
                min_test_metric = rmse
                counter = 0  # Сброс счетчика
            else:
                counter += 1  # Увеличиваем счетчик

            # Проверка условий для ранней остановки
            if counter >= patience or rmse - min_test_metric >= max_delta:
                print(f'Ранняя остановка на эпохе {epoch}')
                break

    return train_loss, test_metric
```

```python
# Функция для кросс-валидации
def cross_val_score(model_class, X, y, num_columns, n_splits=5, num_epochs=10000, lr=5e-5):
    """
    Выполняет кросс-валидацию для заданной модели.

    Параметры:
    - model_class: Класс модели (например, класс нейронной сети).
    - X: Входные данные (матрица признаков).
    - y: Целевые значения (вектор откликов).
    - num_columns: Индексы числовых столбцов для предобработки.
    - n_splits: Количество фолдов для кросс-валидации (по умолчанию 5).
    - num_epochs: Общее количество эпох для обучения модели (по умолчанию 10000).
    - lr: Скорость обучения (по умолчанию 5e-5).
    Возвращает:
    - mean_rmse: Среднее значение RMSE по всем сплитам.
    - all_train_losses: Список потерь для обучения по каждому сплиту.
    - all_test_metrics: Список метрик тестирования по каждому сплиту.
    """

    kf = KFold(n_splits=n_splits, shuffle=True)
    criterion = nn.MSELoss()
    all_rmse_scores = []

    # Определение пайплайна предобработки
    preprocessor = ColumnTransformer(
        transformers=[
        ('num', Pipeline([
          ('imputer', SimpleImputer(strategy='median')),
          ('scaler', MinMaxScaler())
        ]), num_columns),
        ],
    remainder='passthrough'  # Оставляем остальные столбцы без изменений
    )

    all_train_losses = []  # Списки для хранения всех train_loss
    all_test_metrics = []  # Списки для хранения всех test_metrics

    for train_index, val_index in kf.split(X):
        X_train, X_val = X.iloc[train_index], X.iloc[val_index]  # Выборка обучающей и валидационной выборок
        y_train, y_val = y.iloc[train_index], y.iloc[val_index]

        # Предобработка данных
        X_train = preprocessor.fit_transform(X_train)
        X_val = preprocessor.transform(X_val)

        # Перемещение данных на устройство (например, GPU)
        device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
        X_train_tensor = torch.tensor(X_train, dtype=torch.float32).to(device)
        y_train_tensor = torch.tensor(y_train.values.reshape(-1, 1), dtype=torch.float32).to(device)  # Преобразуем целевую переменную
        X_val_tensor = torch.tensor(X_val, dtype=torch.float32).to(device)
        y_val_tensor = torch.tensor(y_val.values.reshape(-1, 1), dtype=torch.float32).to(device)  # Преобразуем целевую переменную

        # Обновляем количество входных нейронов
        n_in_neurons = X_train_tensor.shape[1]
        n_hidden_neurons_1 = 196
        n_hidden_neurons_2 = 148
        n_out_neurons = 1  # Один выход для регрессии

        # Создание экземпляра модели и оптимизатора
        model = model_class(n_in_neurons, n_hidden_neurons_1, n_hidden_neurons_2, n_out_neurons).to(device)
        optimizer = torch.optim.Adamax(model.parameters(), lr=lr)

        # Объявляем списки для хранения данных в этом сплите
        train_loss = []  # Инициализация пустого списка для потерь
        test_metric = []  # Инициализация пустого списка для метрики

        # Обучение модели
        _, test_metric = fit(
            model,
            optimizer,
            criterion,
            X_train_tensor,
            y_train_tensor,
            X_val_tensor,
            y_val_tensor,
            train_loss,
            test_metric,
            num_epochs)

        # Получение RMSE для валидационного набора и добавление в общий список
        all_rmse_scores.append(test_metric[-1])  # Предполагаем, что последнее значение - RMSE для последней эпох
        all_train_losses.append([loss.item() for loss in train_loss])  # Преобразуем в обычные списки
        all_test_metrics.append([metric.item() for metric in test_metric])  # Преобразуем в обычные списки

    mean_rmse = np.mean(all_rmse_scores)  # Среднее RMSE по всем сплитам
    return mean_rmse, all_train_losses, all_test_metrics  # Возврат среднего RMSE и списков потерь и метрик
```

## Тестирование

Тестирование проведём с дообучением модели на полной тренировочной выборке.


```python
# Определение пайплайна предобработки
preprocessor = ColumnTransformer(
    transformers=[
        ('num', Pipeline([
            ('imputer', SimpleImputer(strategy='median')),
            ('scaler', MinMaxScaler())
        ]), num_columns),  # num_columns - индексы числовых столбцов
    ],
    remainder='passthrough'  # Оставляем остальные столбцы без изменений
)

# Настройка устройства выполнения (CPU или GPU)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# Предобработка полных данных
X_train_transformed = preprocessor.fit_transform(X)  # Преобразование обучающих данных
X_test_transformed = preprocessor.transform(X_test)  # Преобразование тестовых данных

# Преобразование данных в тензоры
X_train_tensor = torch.tensor(X_train_transformed, dtype=torch.float32).to(device)
X_test_tensor = torch.tensor(X_test_transformed, dtype=torch.float32).to(device)

y_train_tensor = torch.tensor(y.values.reshape(-1, 1), dtype=torch.float32).to(device)  # Преобразование целевой переменной
y_test_tensor = torch.tensor(y_test.values.reshape(-1, 1), dtype=torch.float32).to(device)  # Преобразование тестовых целевых переменных
y_test_tensor_ = torch.tensor(y_test.values, dtype=torch.float32).to(device)  # Преобразование тестовых целевых переменных без reshape

# Определение экземпляров сетей и количества нейронов
n_in_neurons = X_train_tensor.shape[1]
n_hidden_neurons_1 = 196
n_hidden_neurons_2 = 148
n_out_neurons = 1  # Один выход для регрессии

# Инициализация сети
net_test = Net(
    n_in_neurons,
    n_hidden_neurons_1,
    n_hidden_neurons_2,
    n_out_neurons
)

# Определение оптимизаторов, функции потерь и количества эпох
optimizer = torch.optim.Adamax(net_test.parameters(), lr=5e-2)
loss = nn.MSELoss()  # Функция потерь: MSE
num_epochs = 10000  # Количество эпох

# Создание списков для дальнейшего отображения графиков
train_loss = []  # Список для потерь обучения
test_metric = []  # Список для тестовых метрик
```

```python
# Определение функции для обучения нейронной сети
def fit_test(
    net,
    optimizer,
    train_loss,
    test_metric,
    patience=1,
    max_delta=1
):
    """
    Обучает модель нейронной сети с использованием ранней остановки.

    Параметры:
    - net: Нейронная сеть для обучения.
    - optimizer: Оптимизатор для обновления параметров модели.
    - train_loss: Список для хранения значений потерь на обучающей выборке.
    - test_metric: Список для хранения метрик на валидационной выборке.
    - patience: Количество эпох без улучшения перед остановкой.
    - max_delta: Максимально допустимая разница между текущей и минимальной метриками.
    """

    counter = 0                          # Счетчик эпох, когда валидационная метрика не улучшается
    min_test_metric = np.inf            # Инициализация минимальной тестовой метрики

    # Цикл по эпохам
    for epoch in range(num_epochs):
        optimizer.zero_grad()             # Обнуляем градиенты модели перед новым шагом

        # Получение предсказаний на основе обучающих данных
        preds = net(X_train_tensor).flatten().view(-1, 1)

        # Расчет потерь
        loss_value = loss(preds, y_train_tensor)  # Вычисление потерь
        loss_value.backward()                     # Обратное распространение ошибки

        optimizer.step()                         # Обновление параметров модели

        # Мониторинг потерь и метрик
        if epoch % 10 == 0 or epoch == num_epochs - 1:
            train_loss.append(torch.sqrt(loss_value).detach())  # Добавление потерь в список

            net.eval()                         # Переключаем модель в режим оценки

            # Получение предсказаний на тестовых данных
            test_preds = net(X_test_tensor).flatten().view(-1, 1)

            # Вычисление RMSE на валидационной выборке
            rmse = torch.sqrt(loss(test_preds, y_test_tensor)).detach()
            test_metric.append(rmse)          # Добавляем RMSE в список метрик валидации

            # Ранняя остановка
            if rmse < min_test_metric:
                min_test_metric = rmse         # Обновление минимальной метрики
                counter = 0                     # Сброс счетчика
            else:
                counter += 1                    # Увеличиваем счетчик

            # Проверка условий для ранней остановки
            if counter >= patience or rmse - min_test_metric >= max_delta:
                print(f'Обучение остановлено на эпохе {epoch}')  # Сообщение о ранней остановке
                break
```

```python
# Обучение нейронной сети
fit_test(
    net_test,
    optimizer,
    train_loss,
    test_metric,
    100,
    1
    )
```

```python
# Определение функции для построения графиков
def fit_plot(loss, metric, abb=0):
    """
    Строит графики потерь на тренировочной и тестовой выборках.

    Параметры:
    - loss: Список значений потерь на обучающей выборке.
    - metric: Список метрик на тестовой выборке.
    - abb: Признак для выбора стиля графика (0 - все признаки, 1 - без одного признака).
    """
    # Создание фигуры для графика
    fig = go.Figure()

    # Установка цветов и заголовков графика в зависимости от параметра abb
    if abb == 1:
        fig.add_trace(go.Scatter(
            x=np.array(range(len(loss))) * 10,  # Эпохи для графика потерь
            y=loss,
            name='Тренировка',
            line=dict(color='darkblue', width=3)
        ))
        fig.add_trace(go.Scatter(
            x=np.array(range(len(metric))) * 10,  # Эпохи для графика метрик
            y=metric,
            name='Тест',
            line=dict(color='crimson', width=3)
        ))
        title_text = 'Значения функции потерь при тренировке и метрики на тесте без одного признака'
    else:
        fig.add_trace(go.Scatter(
            x=np.array(range(len(loss))) * 10,  # Эпохи для графика потерь
            y=loss,
            name='Тренировка',
            line=dict(color='darkcyan', width=3)
        ))
        fig.add_trace(go.Scatter(
            x=np.array(range(len(metric))) * 10,  # Эпохи для графика метрик
            y=metric,
            name='Тест',
            line=dict(color='coral', width=3)
        ))
        title_text = 'Значения функции потерь при тренировке и тестировании на тесте со всеми признаками'

    # Обновление макета графика
    fig.update_layout(
        width=1000,
        height=500,
        legend_orientation='h',
        title=dict(
            text=title_text,
            x=0.5
        ),
        xaxis_title='Эпоха',
        yaxis_title='RMSE'
    )

    # Показ графика
    fig.show()
```

```python
# просмотр процесса обучения сетей с большим количеством нейронов
fit_plot(train_loss, test_metric)
```

## Анализ важности признаков:

Оценим важность признаков для лучшей модели и построим графики важности с помощью метода SHAP.

```python
# Преобразование тестовых данных в numpy
X_test_np = X_test_tensor.detach().cpu().numpy()

# Определение функции предсказания
def model_predict(input_data):
    input_tensor = torch.FloatTensor(input_data)
    with torch.no_grad():  # Отключение градиентов для повышения производительности
        predictions = net_test(input_tensor).detach().cpu().numpy()  # Получение предсказаний
    return predictions

# Создаем экземпляр KernelExplainer
explainer = shap.KernelExplainer(model_predict, X_test_np)  # Используем все тестовые данные для обучения

# Вычисление SHAP значений
shap_values = explainer.shap_values(X_test_np, nsamples=100)

# Удаление лишнего измерения
shap_values = np.squeeze(shap_values)

# Проверка новой формы
print("Shape of shap_values after squeeze:", shap_values.shape)

# Создаем объект Explanation с именами признаков
shap_values_obj = shap.Explanation(
    values=shap_values,
    base_values=explainer.expected_value,
    data=X_test_np,  # Используйте numpy массив
    feature_names=X_test.columns.tolist()  # Указываем имена признаков
)
```

```python
# Визуализируем важность признаков
plt.title('График общей значимости признаков нейросети torch.nn')
plt.ylabel('Признаки', fontsize=14)
shap.plots.bar(shap_values_obj)

# Или используем summary plot для более наглядного представления
plt.title('График вклада признаков в каждое предсказание модели:')
plt.ylabel('Признаки', fontsize=14)
shap.summary_plot(shap_values, X_test_np, feature_names=X_test.columns.tolist())
```
